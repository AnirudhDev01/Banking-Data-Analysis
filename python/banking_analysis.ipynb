```python
df = pd.read_csv('../data/Banking_cleaned.csv')
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nCustomer age:")
print(df["Age"].describe())

print("\nEstimated income:")
print(df["Estimated Income"].describe())
```

```python
df["Loyalty Classification"].value_counts()
```

```python
df["Loyalty Classification"].value_counts().plot(kind="bar")

plt.title("Customer Distribution by Loyalty Classification")
plt.xlabel("Loyalty Classification")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# Age distribution
df["Age"].plot(kind="hist", bins=15)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

```python
# Estimated Income distribution
df["Estimated Income"].plot(kind="hist", bins=20)

plt.title("Estimated Income Distribution")
plt.xlabel("Estimated Income")
plt.ylabel("Number of Customers")
plt.show()

plt.title("Estimated Income Distribution")
plt.xlabel("Estimated Income")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df['Risk Weighting'].value_counts().sort_index()
```

```python
df["Risk Weighting"].value_counts().sort_index().plot(kind="bar")

plt.title("Customers by Risk Weighting")
plt.xlabel("Risk Weighting")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df.groupby("Loyalty Classification")["Estimated Income"].mean().round(2)
```

```python
df.groupby("Loyalty Classification")["Estimated Income"].mean().plot(kind="bar")

plt.title("Average Income by Loyalty Classification")
plt.xlabel("Loyalty Classification")
plt.ylabel("Average Income")
plt.show()
```

```python
#Average financial values by loyalty classification
df.groupby("Loyalty Classification")[
    ["Estimated Income", "Bank Deposits", "Bank Loans", "Saving Accounts"]
].mean().round(2)
```

```python
# Average financial value by financial weighting
df.groupby("Risk Weighting")[
    ["Estimated Income", "Bank Deposits", "Bank Loans"]
].mean().round(2)
```

```python
# Top 10 customers by bank deposits
df.nlargest(10, "Bank Deposits")[
    ["Client ID", "Name", "Bank Deposits"]
]
```

```python
# Top 10 customers by bank loans
df.nlargest(10, "Bank Loans")[
    ["Client ID", "Name", "Bank Loans"]
]
```

```python
# Correlation between main financial variables
df[
    ["Estimated Income", "Bank Deposits", "Bank Loans",
     "Saving Accounts", "Checking Accounts", "Credit Card Balance"]
].corr().round(2)
```
